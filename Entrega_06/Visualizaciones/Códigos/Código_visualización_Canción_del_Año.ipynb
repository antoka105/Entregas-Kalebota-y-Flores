```python
import altair as alt
import pandas as pd

# Cargar tu CSV
df = pd.read_csv("Canción del Año.csv")

# Renombrar columnas para asegurar que 'Año' sea accesible
df = df.rename(columns={
    'A�o': 'Año',
    'Categor�a': 'Categoría',
    'Disco o canci�n': 'Disco o canción',
    'G�nero del artista/banda': 'Género del artista/banda',
    'G�nero canci�n': 'Género canción'
})

# Reemplazar valor en columna 'Género disco' (moved out of rename)
df['Género canción'] = df['Género canción'].replace({'Reggeat�n': 'Reggeatón'})
df['Género canción'] = df['Género canción'].replace({'Jazz Pop': 'Subgénero Pop', 'Country Pop':'Subgénero Pop',  'Rock Pop':'Subgénero Pop',  'Folk Pop':'Subgénero Pop',  'Rap Pop':'Subgénero Pop'})

# Agrupar por año y género del disco
genre_year = (
    df.groupby(["Año", "Género canción"])
      .size()
      .reset_index(name="Cantidad")
)

# Crear gráfico de áreas apiladas por género
chart = (
    alt.Chart(genre_year)
    .mark_area()
    .encode(
        x=alt.X("Año:O").title("Año"),
        y=alt.Y("sum(Cantidad):Q")
            .stack("center")
            .title("Cantidad de nominaciones"),
        color=alt.Color("Género canción:N").title("Género"),
        tooltip=["Año", "Género canción", "Cantidad"]
    )
    .interactive()
)

chart
```

```python
import altair as alt
import pandas as pd

# Cargar CSV
df = pd.read_csv("Canción del Año.csv")

# Renombrar columnas para asegurar que 'Año' sea accesible
df = df.rename(columns={
    'A�o': 'Año',
    'Categor�a': 'Categoría',
    'Disco o canci�n': 'Disco o canción',
    'G�nero del artista/banda': 'Género del artista/banda',
    'G�nero canci�n': 'Género canción'
})

# Reemplazar valor en columna 'Género disco' (moved out of rename)
df['Género canción'] = df['Género canción'].replace({'Reggeat�n': 'Reggeatón'})
df['Género canción'] = df['Género canción'].replace({'Jazz Pop': 'Subgénero Pop', 'Country Pop':'Subgénero Pop',  'Rock Pop':'Subgénero Pop',  'Folk Pop':'Subgénero Pop',  'Rap Pop':'Subgénero Pop', 'Latin Pop':'Subgénero Pop'})

# Agrupar datos
genre_year = (
    df.groupby(["Año", "Género canción"])
      .size()
      .reset_index(name="Cantidad")
)

scale=alt.Scale(
                domain=["Pop", "Rock", "Jazz", "Alternativo", "Country", "Folk", "R&B", "Rap", "Trap", "Americana", "House", "Disco", "Blues", "Soul", "Subgénero Pop"],  # <- géneros presentes
                range=["#EFCE7B", "#252D45", "#D3B6D3", "#FF7BAC", "#25533F", "#6D1F42", '#B8CEE8', '#EF6F3C', '#AACC96', '#F4BEAE', '#52A5CE','#876029', '#AFAB23', '#6777B6', '#EA5E86'  ]  # <- colores elegidos
)
chart = (
    alt.Chart(genre_year)
    .mark_area()
    .encode(
        x=alt.X("Año:O").title("Año"),
        y=alt.Y("sum(Cantidad):Q")
            .stack("center")
            .title("Cantidad de nominaciones"),
        color=alt.Color("Género canción:N")
            .title("Género")
            .scale(scale)
            .sort(alt.EncodingSortField(field="Cantidad", op="sum", order="descending")), # Apply sort to color
        tooltip=["Año", "Género canción", "Cantidad"]
    )
    .interactive()
)
chart.save('chart.html')
chart
```

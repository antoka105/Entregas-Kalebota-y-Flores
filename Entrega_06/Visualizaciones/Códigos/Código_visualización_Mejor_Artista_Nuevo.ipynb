```python
import altair as alt
import pandas as pd

# Cargar tu CSV
df = pd.read_csv("/content/Mejor Artista Nuevo.csv")

# Renombrar columnas para asegurar que 'Año' sea accesible
df = df.rename(columns={
    'A�o': 'Año',
    'Categor�a': 'Categoría',
    'G�nero del artista/banda': 'Género del artista/banda',
})

# Agrupar por año y género del disco
genre_year = (
    df.groupby(["Año", "Género del artista/banda"])
      .size()
      .reset_index(name="Cantidad")
)

# Crear gráfico de áreas apiladas por género
chart = (
    alt.Chart(genre_year)
    .mark_area()
    .encode(
        x=alt.X("Año:O").title("Año"),
        y=alt.Y("sum(Cantidad):Q")
            .stack("center")
            .title("Cantidad de nominaciones"),
        color=alt.Color("Género del artista/banda:N").title("Género"),
        tooltip=["Año", "Género del artista/banda", "Cantidad"]
    )
    .interactive()
)

chart
```

```python
import altair as alt
import pandas as pd

# Cargar tu CSV
df = pd.read_csv("/content/Mejor Artista Nuevo.csv")

# Renombrar columnas para asegurar que 'Año' sea accesible
df = df.rename(columns={
    'A�o': 'Año',
    'Categor�a': 'Categoría',
    'G�nero del artista/banda': 'Género del artista/banda',
})

# Agrupar por año y género del disco
genre_year = (
    df.groupby(["Año", "Género del artista/banda"])
      .size()
      .reset_index(name="Cantidad")
)

# Escala de colores personalizada
scale = alt.Scale(
    domain=["H","M","MIXTO"],
    range=["#EFCE7B", "#252D45", "#B8CEE8"]
)

# Crear gráfico de áreas apiladas
chart = (
    alt.Chart(genre_year)
    .mark_area()
    .encode(
        x=alt.X("Año:O").title("Año"),
        y=alt.Y("sum(Cantidad):Q")
              .stack("center")
              .title("Cantidad de nominaciones"),
        color=alt.Color(
            "Género del artista/banda:N",
            title="Género",
            scale=scale
        ),
        tooltip=["Año", "Género del artista/banda", "Cantidad"]
    )
    .interactive()
)

chart
```

```python
import altair as alt
import pandas as pd

# Cargar tu CSV
df = pd.read_csv("Álbum del Año.csv")

# Renombrar columnas para asegurar que 'Año' sea accesible
df = df.rename(columns={
    'A�o': 'Año',
    'Categor�a': 'Categoría',
    'Disco o canci�n': 'Disco o canción',
    'G�nero del arista/banda': 'Género del arista/banda',
    'G�nero disco': 'Género disco' })

# Reemplazar valor en columna 'Género disco' (moved out of rename)
df['Género disco'] = df['Género disco'].replace({'Reggeat�n': 'Reggeatón'})
df['Género disco'] = df['Género disco'].replace({'Dance Pop': 'Subgénero Pop', 'Country Pop':'Subgénero Pop',  'Pop Rock':'Subgénero Pop'})


# Agrupar por año y género del disco
genre_year = (
    df.groupby(["Año", "Género disco"])
      .size()
      .reset_index(name="Cantidad")
)

# Crear gráfico de áreas apiladas por género
chart = (
    alt.Chart(genre_year)
    .mark_area()
    .encode(
        x=alt.X("Año:O").title("Año"),
        y=alt.Y("sum(Cantidad):Q")
            .stack("center")
            .title("Cantidad de nominaciones"),
        color=alt.Color("Género disco:N").title("Género"),
        tooltip=["Año", "Género disco", "Cantidad"]
    )
    .interactive()
)

chart
```

```python
import altair as alt
import pandas as pd

# Cargar CSV
df = pd.read_csv("Álbum del Año.csv")

# Renombrar columnas para asegurar que 'Año' sea accesible
df = df.rename(columns={
    'A�o': 'Año',
    'Categor�a': 'Categoría',
    'Disco o canci�n': 'Disco o canción',
    'G�nero del arista/banda': 'Género del arista/banda',
    'G�nero disco': 'Género disco'
})

# Reemplazar valor en columna 'Género disco' (moved out of rename)
df['Género disco'] = df['Género disco'].replace({'Reggeat�n': 'Reggeatón'})
df['Género disco'] = df['Género disco'].replace({'Dance Pop': 'Subgénero Pop', 'Country Pop':'Subgénero Pop',  'Pop Rock':'Subgénero Pop'})

# Agrupar datos
genre_year = (
    df.groupby(["Año", "Género disco"])
      .size()
      .reset_index(name="Cantidad")
)

scale=alt.Scale(
                domain=["Pop", "Rock", "Jazz", "Alternativo", "Country", "Folk", "R&B", "Rap", "Afrobeat", "Americana", "Banda Sonora", "Disco", "Reggeatón", "Soul", "Subgénero Pop"],  # <- géneros presentes
                range=["#EAC119", "#808BC5", "#EAA7C7", "#9ED6DF", "#245E55", "#ED773C", '#C63F3E', '#1D1D1B', '#D8639C', '#EDD470', '#49B5A3','#444E7D', '#24B2C9', '#DA7676', '#F1A781'  ]  # <- colores elegidos
)

chart = (
    alt.Chart(genre_year)
    .mark_area()
    .encode(
        x=alt.X("Año:O").title("Año"),
        y=alt.Y("sum(Cantidad):Q")
            .stack("center")
            .title("Cantidad de nominaciones"),
        color=alt.Color("Género disco:N")
            .title("Género")
            .scale(scale)
            .sort(alt.EncodingSortField(field="Cantidad", op="sum", order="descending")), # Apply sort to color
        tooltip=["Año", "Género disco", "Cantidad"]
    )
    .interactive()
)
chart.save('chart.html')
chart
```

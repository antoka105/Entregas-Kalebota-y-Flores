```python
import pandas as pd
import altair as alt

# Cargar datos
df = pd.read_csv("/content/Álbum del Año.csv")

# Renombrar columnas
df = df.rename(columns={
    'A�o': 'Año',
    'Categor�a': 'Categoría',
    'Disco o canci�n': 'Disco o canción',
    'G�nero del arista/banda': 'Género del arista/banda',
    'G�nero disco': 'Género disco'
})

# Reemplazar valor en columna 'Género disco'
df['Género disco'] = df['Género disco'].replace({'Reggeat�n': 'Reggeatón'})

df.head(8)
```

```python
# Agrupar por año y género del disco
genero_anual = (
    df.groupby(["Año", "Género disco"])
    .size()
    .reset_index(name="Cantidad")
)

# Crear visualización con barras horizontales
chart = (
    alt.Chart(genero_anual)
    .mark_bar()
    .encode(
        y=alt.Y("Año:O", title="Año", sort='-x'),
        x=alt.X(
            "Cantidad:Q",
            title="Cantidad de artistas"
        ),
        color=alt.Color("Género disco:N", title="Género musical"),
        tooltip=["Año", "Género disco", "Cantidad"]
    )
    .properties(
        title="Cantidad de artistas por género musical y año",
        width=600,
        height=400
    )
)

chart
```

```python
# Agrupar por año y género
genero_anual = (
    df.groupby(["Año", "Género disco"])
    .size()
    .reset_index(name="Cantidad")
)

# Asegurar que Cantidad sea entero
genero_anual["Cantidad"] = genero_anual["Cantidad"].astype(int)

# Forzar ticks enteros y formato 'd' (enteros)
chart = (
    alt.Chart(genero_anual)
    .mark_bar()
    .encode(
        y=alt.Y("Año:O", title="Año"),
        x=alt.X(
            "Cantidad:Q",
            title="Cantidad de artistas",
            axis=alt.Axis(tickMinStep=1, format='d')
        ),
        color=alt.Color("Género disco:N", title="Género musical"),
        tooltip=["Año", "Género disco", "Cantidad"]
    )
    .properties(width=700, height=400, title="Cantidad de artistas por género musical y año")
)

chart
```
